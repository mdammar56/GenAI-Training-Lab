```python
from google.colab import userdata

GROQ_API_KEY = userdata.get("GROQ_API_KEY")

if not GROQ_API_KEY:
    raise ValueError("Please add GROQ_API_KEY to your Colab Secrets.")

print("Groq API key loaded successfully!")
```

```python
from google.colab import files

uploaded = files.upload()

pdf_files = [
    name for name in uploaded.keys()
    if name.lower().endswith(".pdf")
]

if not pdf_files:
    raise ValueError("Please upload a PDF file.")

file_name = pdf_files[0]

print("PDF uploaded successfully:", file_name)
```

```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader(file_name)

documents = loader.load()

print("PDF loaded successfully!")
print("Total pages:", len(documents))

for i, doc in enumerate(documents[:3]):
    print(f"\n--- Page {i + 1} ---")
    print(doc.page_content[:500])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100
)

chunks = splitter.split_documents(documents)

print("PDF split successfully!")
print("Total chunks created:", len(chunks))

for i, chunk in enumerate(chunks[:3]):
    print(f"\n--- Chunk {i + 1} ---")
    print(chunk.page_content[:500])
    print("Page:", chunk.metadata.get("page", "Unknown"))
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

print("Embedding model loaded successfully!")
```

```python
from langchain_chroma import Chroma

vectorstore = Chroma.from_documents(
    documents=chunks,
    embedding=embeddings,
    collection_name="pdf_rag_collection"
)

print("ChromaDB vector store created successfully!")
print("Total chunks stored:", len(chunks))
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    api_key=GROQ_API_KEY,
    model="openai/gpt-oss-20b",
    temperature=0
)

print("Groq language model initialized successfully!")
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough

# Create the retriever
retriever = vectorstore.as_retriever(
    search_kwargs={"k": 4}
)

# Format retrieved PDF sections
def format_docs(docs):
    return "\n\n".join(
        f"Page {doc.metadata.get('page', 0) + 1}:\n{doc.page_content}"
        for doc in docs
    )

# Create the prompt
rag_prompt = ChatPromptTemplate.from_template("""
You are a helpful PDF document assistant.

Answer the user's question using only the information
provided in the document context.

Instructions:
- Give clear, accurate, and well-organized answers.
- Use simple language whenever possible.
- If the answer is not available in the context,
  say: "I couldn't find the answer in the uploaded PDF."
- Do not invent information.
- Mention page numbers when useful.

Document Context:
{context}

User Question:
{question}

Answer:
""")

# Build the RAG chain
rag_chain = (
    {
        "context": retriever | format_docs,
        "question": RunnablePassthrough()
    }
    | rag_prompt
    | llm
    | StrOutputParser()
)

print("PDF RAG chain created successfully!")
```

```python
question = "Summarize the uploaded PDF."

answer = rag_chain.invoke(question)

print("Question:", question)
print("\nAnswer:")
print(answer)
```

```python

import gradio as gr
from pypdf import PdfReader
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_chroma import Chroma
import tempfile
import os
import shutil
import uuid

# Store the active PDF's RAG components
rag_data = {
    "chain": None,
    "name": None,
    "pages": 0,
    "chunks": 0,
    "db": None
}


# Process uploaded PDF
def process_pdf(pdf_file):
    if pdf_file is None:
        return "⚠️ Please upload a PDF first.", []

    try:
        pdf_path = pdf_file.name

        # Load PDF pages
        loader = PyPDFLoader(pdf_path)
        documents = loader.load()

        if not documents or not any(
            doc.page_content.strip() for doc in documents
        ):
            return "⚠️ No readable text found. Try a text-based PDF.", []

        # Split into chunks
        splitter = RecursiveCharacterTextSplitter(
            chunk_size=500,
            chunk_overlap=100
        )
        chunks = splitter.split_documents(documents)

        if not chunks:
            return "⚠️ No text chunks could be created.", []

        # Create a fresh vector store for this PDF
        collection_name = "pdf_" + uuid.uuid4().hex[:12]

        vectorstore = Chroma.from_documents(
            documents=chunks,
            embedding=embeddings,
            collection_name=collection_name
        )

        # Create retriever
        retriever = vectorstore.as_retriever(
            search_kwargs={"k": 4}
        )

        # Format retrieved text
        def format_docs(docs):
            return "\n\n".join(
                f"Page {doc.metadata.get('page', 0) + 1}:\n"
                f"{doc.page_content}"
                for doc in docs
            )

        # Prompt
        prompt = ChatPromptTemplate.from_template("""
You are a helpful PDF assistant.

Answer using only the retrieved context.
If the answer is not present, say:
"I couldn't find the answer in the uploaded PDF."

Give clear, accurate answers.
Mention page numbers when useful.

Context:
{context}

Question:
{question}

Answer:
""")

        # Create RAG chain
        chain = (
            {
                "context": retriever | format_docs,
                "question": RunnablePassthrough()
            }
            | prompt
            | llm
            | StrOutputParser()
        )

        # Save current PDF's components
        rag_data["chain"] = chain
        rag_data["name"] = os.path.basename(pdf_path)
        rag_data["pages"] = len(documents)
        rag_data["chunks"] = len(chunks)
        rag_data["db"] = vectorstore

        status = f"""
✅ PDF processed successfully!

📄 **File:** {rag_data['name']}

📑 **Pages:** {len(documents)}

🧠 **Text chunks:** {len(chunks)}

You can now ask questions about your document.
"""

        return status, []

    except Exception as e:
        return f"❌ Error processing PDF: {type(e).__name__}: {str(e)}", []


# Answer user questions
def chat_with_pdf(question, history):
    history = history or []

    if not question or not question.strip():
        history.append({
            "role": "assistant",
            "content": "Please enter a question."
        })
        return history, ""

    if rag_data["chain"] is None:
        history.append({
            "role": "user",
            "content": question
        })
        history.append({
            "role": "assistant",
            "content": "Please upload and process a PDF first."
        })
        return history, ""

    history.append({
        "role": "user",
        "content": question
    })

    try:
        answer = rag_data["chain"].invoke(question.strip())
    except Exception as e:
        answer = f"Error: {type(e).__name__}: {str(e)}"

    history.append({
        "role": "assistant",
        "content": answer
    })

    return history, ""


# Professional styling
custom_css = """
.gradio-container {
    max-width: 1150px !important;
    margin: auto !important;
}

#main-title {
    text-align: center;
    padding: 25px 10px;
}

#main-title h1 {
    font-size: 36px;
    font-weight: 800;
}

#main-title p {
    font-size: 16px;
    opacity: 0.8;
}

#upload-panel, #chat-panel {
    border-radius: 16px;
    padding: 15px;
}
"""


# Build interface
with gr.Blocks() as app:

    gr.HTML("""
    <div id="main-title">
        <h1>✦ PDF Intelligence</h1>
        <p>Upload a document. Ask questions. Discover insights.</p>
    </div>
    """)

    with gr.Row():

        with gr.Column(scale=1, elem_id="upload-panel"):

            gr.Markdown("## 📄 Upload your PDF")

            pdf_upload = gr.File(
                label="Choose a PDF document",
                file_types=[".pdf"],
                type="filepath"
            )

            process_button = gr.Button(
                "⚡ Process PDF",
                variant="primary"
            )

            status_box = gr.Markdown(
                "Upload a PDF and click **Process PDF** to begin."
            )

            gr.Markdown("---")

            gr.Markdown("""
### ✨ Example questions

- Summarize this document.
- What are the main topics?
- Explain the important points.
- What is the conclusion?
- Find specific information in the PDF.
""")

            gr.Markdown("---")

            gr.Markdown("""
**Technology**

🧠 Groq AI
🔗 LangChain
🔎 ChromaDB
🤗 Hugging Face
""")

        with gr.Column(scale=2, elem_id="chat-panel"):

            gr.Markdown("## 💬 Chat with your PDF")

            chatbot = gr.Chatbot(
                label="Conversation",
                height=430
            )

            question_box = gr.Textbox(
                label="Your question",
                placeholder="Ask anything about your processed PDF...",
                lines=2
            )

            with gr.Row():

                send_button = gr.Button(
                    "✦ Ask AI",
                    variant="primary"
                )

                clear_button = gr.Button("Clear Chat")


    # Process PDF when button is clicked
    process_button.click(
        fn=process_pdf,
        inputs=[pdf_upload],
        outputs=[status_box, chatbot]
    )

    # Ask questions
    send_button.click(
        fn=chat_with_pdf,
        inputs=[question_box, chatbot],
        outputs=[chatbot, question_box]
    )

    question_box.submit(
        fn=chat_with_pdf,
        inputs=[question_box, chatbot],
        outputs=[chatbot, question_box]
    )

    # Clear conversation
    clear_button.click(
        fn=lambda: ([], ""),
        inputs=[],
        outputs=[chatbot, question_box]
    )


app.launch(
    share=True,
    theme=gr.themes.Soft(
        primary_hue="violet",
        secondary_hue="blue",
        neutral_hue="slate"
    ),
    css=custom_css
)
```
